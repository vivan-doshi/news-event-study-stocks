```python
def get_mlr_equation_string(results):
    params = results.params
    intercept = params.get("const", 0.0)

    terms = []
    for name, val in params.items():
        if name == "const":
            continue
        terms.append(f"{val:.4f} * {name}")

    if terms:
        eq = "y = " + f"{intercept:.4f} + " + " + ".join(terms)
    else:
        eq = "y = " + f"{intercept:.4f}"
    return eq

mlr_equation_str = get_mlr_equation_string(results)
print(mlr_equation_str)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

def plot_feature_vs_target_with_original_mlr_line(X, y, results, feature_name, symbol, equation_str):
    """
    For a given feature:
      - Scatter: feature vs actual target (y)
      - Line: predictions from the ORIGINAL multivariate regression,
              varying this feature, holding all other features at their mean.
      - The SAME multivariate equation string is shown on every plot.
    """
    if feature_name not in X.columns:
        print(f"[{symbol}] Feature {feature_name} not in X.")
        return

    # 1) Subset for scatter (actual data)
    sub = X[[feature_name]].copy()
    sub["y_true"] = y
    sub = sub.dropna()

    if sub[feature_name].nunique() < 2:
        print(f"[{symbol} | {feature_name}] Feature nearly constant; skipping.")
        return

    x = sub[feature_name].values
    y_true = sub["y_true"].values

    # 2) Build line using ORIGINAL multivariate model (no refit!)
    params = results.params
    exog_names = results.model.exog_names  # e.g. ['const', 'feat1', 'feat2', ...]

    # Model feature names (exclude const)
    model_features = [name for name in exog_names if name != "const"]

    # keep only model features that exist in X
    model_features_in_X = [f for f in model_features if f in X.columns]

    # Base means from the sample we're plotting on
    X_sample = X.loc[sub.index, model_features_in_X].copy()
    base_means = X_sample.mean()

    # Grid over this feature
    x_grid = np.linspace(x.min(), x.max(), 100)

    rows = []
    for val in x_grid:
        row = base_means.copy()
        row[feature_name] = val
        rows.append(row)

    # DataFrame of features (only those we have)
    X_grid = pd.DataFrame(rows, columns=model_features_in_X)

    # Add any model features that aren't in X_grid as zeros (or some default)
    for feat in model_features:
        if feat not in X_grid.columns:
            X_grid[feat] = 0.0  # default; you can change to a global mean if you want

    # Ensure column order matches model's expectation (excluding const for now)
    X_grid = X_grid[[f for f in exog_names if f != "const"]]

    # Add constant and align exactly with exog_names
    X_grid_const = sm.add_constant(X_grid, has_constant="add")

    # Make sure all exog_names exist; if not, add zero columns
    for name in exog_names:
        if name not in X_grid_const.columns:
            X_grid_const[name] = 0.0

    # Final column order: exactly exog_names
    X_grid_const = X_grid_const[exog_names]

    # Predict using ORIGINAL multivariate model
    y_line = results.predict(X_grid_const)

    # 3) Plot
    plt.figure(figsize=(8, 5))
    plt.scatter(x, y_true, alpha=0.5, label="Actual target")
    plt.plot(x_grid, y_line, linestyle="--", label="Original MLR line")

    plt.xlabel(feature_name)
    plt.ylabel(y.name or "target")
    plt.title(f"{symbol}: {y.name or 'target'} vs {feature_name}")

    # EXACT SAME equation string on all plots
    '''
    plt.figtext(
        0.5, -0.05,
        equation_str,
        wrap=True,
        horizontalalignment="center",
        fontsize=9
    )
    '''

    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
def prepare_symbol_data(symbol, cutoff_date='2025-06-01'):
    """
    Mirror your apple_df / train-test logic for any symbol.
    Uses global avg_sentiment_df, features, and target.
    """
    df_sym = avg_sentiment_df.loc[avg_sentiment_df['symbol_query'] == symbol, :].copy()
    df_sym = df_sym.sort_values('date')

    # Drop rows where target or any feature is NaN
    df_sym = df_sym.dropna(subset=features + [target])

    df_train = df_sym.loc[df_sym['date'] < cutoff_date, :].copy()
    df_test  = df_sym.loc[df_sym['date'] >= cutoff_date, :].copy()

    X_train = df_train[features]
    y_train = df_train[target]

    X_test  = df_test[features]
    y_test  = df_test[target]

    return X_train, y_train, X_test, y_test, df_sym
```

```python
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

def run_ols_for_symbol(symbol, cutoff_date='2025-06-01', make_plots=True):
    X_train, y_train, X_test, y_test, df_sym = prepare_symbol_data(symbol, cutoff_date)

    # Add constant and fit model (same as before)
    X_train_const = sm.add_constant(X_train)
    model = sm.OLS(y_train, X_train_const)
    results = model.fit()

    # Predictions on test set
    X_test_const = sm.add_constant(X_test)
    y_pred = results.predict(X_test_const)

    # Residuals
    residuals = y_test - y_pred

    if make_plots:
        # (1) Residuals vs predicted (unchanged)
        plt.figure(figsize=(10, 6))
        plt.scatter(y_pred, residuals, alpha=0.5)
        plt.axhline(y=0, color='r', linestyle='--')
        plt.xlabel('Predicted Values')
        plt.ylabel('Residuals')
        plt.title(f'{symbol} – Residuals vs. Predicted Values (OLS)')
        plt.grid(True)
        plt.show()

        # (2) Actual vs predicted with *regression line through the points*
        plt.figure(figsize=(10, 6))
        plt.scatter(y_test, y_pred, alpha=0.5, label='Observed points')

        # Fit simple regression: y_pred = a + b * y_test
        coeffs = np.polyfit(y_test, y_pred, 1)  # [slope, intercept]
        slope, intercept = coeffs[0], coeffs[1]

        x_line = np.linspace(y_test.min(), y_test.max(), 100)
        y_line = intercept + slope * x_line

        plt.plot(x_line, y_line, linestyle='--', label=f'Regression line: y = {intercept:.2f} + {slope:.2f}x')

        plt.xlabel('Actual Next-Day Close')
        plt.ylabel('Predicted Next-Day Close')
        plt.title(f'{symbol} – Actual vs Predicted (with regression line)')
        plt.legend()
        plt.grid(True)
        plt.show()

    return results, y_test, y_pred, residuals
```

```python
all_results = {}
all_predictions = []

for symbol in MAG7:
    print("=" * 80)
    print(f"Running OLS for {symbol}")
    print("=" * 80)

    results, y_test, y_pred, residuals = run_ols_for_symbol(symbol)

    # Store the statsmodels results object
    all_results[symbol] = results

    # Build a small DataFrame with predictions for that symbol
    df_sym = avg_sentiment_df.loc[avg_sentiment_df['symbol_query'] == symbol, :].copy()
    df_sym = df_sym.sort_values('date')
    df_sym = df_sym[df_sym['date'] >= '2025-06-01'].copy()
    df_sym = df_sym.loc[df_sym[target].notna(), :]

    df_out = pd.DataFrame({
        'symbol_query': symbol,
        'date': df_sym['date'].values[:len(y_test)],
        'actual_next_day_close': y_test.values,
        'predicted_next_day_close': y_pred.values,
        'residual': residuals.values
    })

    all_predictions.append(df_out)

# Combine predictions across all MAG7
predictions_mag7 = pd.concat(all_predictions, ignore_index=True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

def plot_scatter_with_regression(df, x_col, y_col, title=None):
    # Keep only the two columns and drop rows with NaN
    sub = df[[x_col, y_col]].dropna().copy()
    x = sub[x_col].values
    y = sub[y_col].values

    # Fit line: y = a + b x using statsmodels (consistent with your OLS)
    X = sm.add_constant(x)          # adds intercept term
    model = sm.OLS(y, X).fit()
    a, b = model.params             # intercept, slope

    # Scatter
    plt.figure(figsize=(8, 5))
    plt.scatter(x, y, alpha=0.5, label="Data points")

    # Regression line
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = a + b * x_line
    plt.plot(x_line, y_line, linestyle="--",
             label=f"Regression line: y = {a:.2f} + {b:.2f}x")

    plt.xlabel(x_col)
    plt.ylabel(y_col)
    if title is None:
        title = f"{y_col} vs {x_col}"
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()

    return model
```
